## Data Source & Identifiers
| Column              | Meaning                                                                                                    |
| ------------------- | ---------------------------------------------------------------------------------------------------------- |
| **data_provider_1** | Primary organization that supplied the system record (utility, state program, installer database, etc.).   |
| **data_provider_2** | Secondary provider of the same record (if applicable).                                                     |
| **system_ID_1**     | Unique system identifier assigned by the primary data provider.                                            |
| **system_ID_2**     | Secondary identifier from another provider or program database.                                            |
| **TTS_link_ID**     | Internal Berkeley Lab ID used to link records across multiple sources within the Tracking the Sun dataset. |


## Installation Characteristics

| Column                       | Meaning                                                                             |
| ---------------------------- | ----------------------------------------------------------------------------------- |
| **installation_date**        | Date the solar PV system was installed or interconnected to the grid.               |
| **PV_system_size_DC**        | Total direct-current (DC) capacity of the PV system (usually in **kWdc**).          |
| **customer_segment**         | Market sector of the installation (e.g., residential, commercial, non-residential). |
| **expansion_system**         | Indicates if the system expanded a previously installed PV system.                  |
| **multiple_phase_system**    | Indicates if the installation occurred in multiple phases.                          |
| **new_construction**         | Whether the PV system was installed on a newly constructed building.                |
| **technology_type**          | Type of solar technology used (e.g., crystalline silicon, thin film).               |
| **extensions_multiphase_id** | Identifier linking multiple system phases or expansions of the same project.        |


## Cost & Incentives
| Column                    | Meaning                                                                   |
| ------------------------- | ------------------------------------------------------------------------- |
| **total_installed_price** | Total cost of the PV system before incentives (typically USD).            |
| **rebate_or_grant**       | Amount of rebate, grant, or incentive applied to the system installation. |
| **battery_price**         | Cost associated with battery storage if included.                         |


## Location Information

| Column                        | Meaning                                             |
| ----------------------------- | --------------------------------------------------- |
| **zip_code**                  | ZIP code where the PV system is located.            |
| **city**                      | City of installation.                               |
| **state**                     | U.S. state where the system is installed.           |
| **utility_service_territory** | Electric utility serving the installation location. |


## Ownership and Installer Info

| Column                | Meaning                                                                                      |
| --------------------- | -------------------------------------------------------------------------------------------- |
| **third_party_owned** | Indicates if the system is owned by a third party (e.g., lease or power purchase agreement). |
| **installer_name**    | Company or contractor that installed the system.                                             |
| **self_installed**    | Indicates if the system was installed by the property owner rather than a contractor.        |


## Mounting & Orientation

| Column                                | Meaning                                                                                                                   |
| ------------------------------------- | ------------------------------------------------------------------------------------------------------------------------- |
| **tracking**                          | Indicates whether the panels use a tracking system to follow the sun.                                                     |
| **ground_mounted**                    | Indicates if the PV system is installed on the ground rather than on a building.                                          |
| **azimuth_1 / azimuth_2 / azimuth_3** | Orientation of module arrays relative to north (degrees). Multiple fields allow systems with multiple array orientations. |
| **tilt_1 / tilt_2 / tilt_3**          | Tilt angle of PV panels from horizontal (degrees).                                                                        |


## PV Module Information
| Column                                  | Meaning                                                                 |
| --------------------------------------- | ----------------------------------------------------------------------- |
| **module_manufacturer_1 / 2 / 3**       | Manufacturer of PV modules used in the system.                          |
| **module_model_1 / 2 / 3**              | Model name of the PV module.                                            |
| **module_quantity_1 / 2 / 3**           | Number of modules of that type installed.                               |
| **additional_modules**                  | Indicates if additional module types exist beyond those listed.         |
| **technology_module_1 / 2 / 3**         | Technology type of the module (e.g., mono-Si, poly-Si, thin film).      |
| **BIPV_module_1 / 2 / 3**               | Indicates whether modules are building-integrated photovoltaics (BIPV). |
| **bifacial_module_1 / 2 / 3**           | Indicates if the module collects light on both sides.                   |
| **nameplate_capacity_module_1 / 2 / 3** | Rated power output of each module type (W).                             |
| **efficiency_module_1 / 2 / 3**         | Conversion efficiency of each module type (%).                          |


## Inverter Information

| Column                                 | Meaning                                                                      |
| -------------------------------------- | ---------------------------------------------------------------------------- |
| **inverter_manufacturer_1 / 2 / 3**    | Manufacturer of the inverter.                                                |
| **inverter_model_1 / 2 / 3**           | Model of the inverter used.                                                  |
| **inverter_quantity_1 / 2 / 3**        | Number of inverters installed of that type.                                  |
| **additional_inverters**               | Indicates if additional inverter types are present beyond those listed.      |
| **micro_inverter_1 / 2 / 3**           | Indicates if the inverter is a micro-inverter attached to individual panels. |
| **built_in_meter_inverter_1 / 2 / 3**  | Indicates if the inverter has an integrated meter.                           |
| **output_capacity_inverter_1 / 2 / 3** | Maximum AC output capacity of the inverter (usually kW).                     |
| **DC_optimizer**                       | Indicates presence of DC power optimizers.                                   |
| **inverter_loading_ratio**             | Ratio of total DC panel capacity to inverter AC capacity (DC/AC ratio).      |



## Battery Storage Information

| Column                         | Meaning                                       |
| ------------------------------ | --------------------------------------------- |
| **battery_manufacturer**       | Manufacturer of the battery storage system.   |
| **battery_model**              | Model of the battery installed.               |
| **battery_rated_capacity_kW**  | Maximum power output of the battery (kW).     |
| **battery_rated_capacity_kWh** | Energy storage capacity of the battery (kWh). |


## Summary of Data Structure

| Category                 | Example Fields                       |
| ------------------------ | ------------------------------------ |
| System metadata          | system_ID, data_provider             |
| Installation details     | installation_date, PV_system_size_DC |
| Financial data           | installed price, rebates             |
| Equipment specifications | modules, inverters, batteries        |
| Location & ownership     | zip_code, utility, third_party_owned |
