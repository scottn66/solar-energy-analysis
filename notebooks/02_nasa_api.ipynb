## Setup

```python
import osimport requestsimport numpy as npimport pandas as pd
```

```python

url = "https://power.larc.nasa.gov/api/temporal/hourly/point"

params = {
    "parameters": "ALLSKY_SFC_SW_DWN,T2M",
    "community": "SB",
    "longitude": -121.8863,
    "latitude": 37.33,
    "start": "20240101",
    "end": "20240131",
    "format": "JSON"
}

response = requests.get(url, params=params)
nasa_daily_data = response.json()
```

This request retrieves hourly solar radiation and temperature data for San Jose during January 2024.

| Parameter         | Meaning                         | Units |
| ----------------- | ------------------------------- | ----- |
| ALLSKY_SFC_SW_DWN | Solar radiation reaching ground | W/m²  |
| T2M               | Temperature at 2 meters         | °C    |
| WS10M             | Wind speed at 10 meters         | m/s   |
| RH2M              | Relative humidity               | %     |
| CLRSKY_SFC_SW_DWN | Clear-sky solar radiation       | W/m²  |


| Community | Use Case                                        |
| --------- | ----------------------------------------------- |
| SB        | Sustainable Buildings (solar / energy modeling) |
| RE        | Renewable energy                                |
| AG        | Agriculture                                     |
| CL        | Climate analysis                                |
