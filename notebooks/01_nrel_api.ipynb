```python
import requests
from google.colab import userdata

nrel = userdata.get('nrel')

url = "https://developer.nrel.gov/api/pvwatts/v8.json"

params = {
    "api_key": nrel,
    "lat": 34.05,
    "lon": -118.25,
    "system_capacity": 5,
    "azimuth": 180,
    "tilt": 20,
    "array_type": 1,
    "module_type": 0,
    "losses": 14
}

response = requests.get(url, params=params)
data = response.json()
```

## What data NREL provides
* hourly or monthly energy generation
* solar radiation
* temperature
* system capacity
* AC/DC power estimates
