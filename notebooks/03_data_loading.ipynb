```python
import os
import pandas as pd
import gdown
import zipfile

FILE_ID = "1NQh4TRC_IqDz2r5vfZuxDm6LGjEuexdu"
OUTPUT = "tracking_the_sun.zip" # Change output name to reflect it's a zip
CSV_FILE_NAME = "TTS_LBNL_public_file_29-Sep-2025_all.csv"

if not os.path.exists(OUTPUT):
    url = f"https://drive.google.com/uc?id={FILE_ID}"
    gdown.download(url, OUTPUT, quiet=False)

# Columns commonly used in analysis
analysis_columns = [
    "installation_date",
    "PV_system_size_DC",
    "total_installed_price",
    "rebate_or_grant",
    "customer_segment",
    "state",
    "zip_code",
    "utility_service_territory",
    "third_party_owned",
    "installer_name",
    "tracking",
    "ground_mounted",
    "azimuth_1",
    "tilt_1",
    "module_manufacturer_1",
    "module_model_1",
    "module_quantity_1",
    "technology_module_1",
    "efficiency_module_1",
    "inverter_manufacturer_1",
    "inverter_model_1",
    "output_capacity_inverter_1",
    "inverter_loading_ratio",
    "battery_rated_capacity_kWh"
]

# Unzip the file
with zipfile.ZipFile(OUTPUT, 'r') as zip_ref:
    zip_ref.extractall('.')

# Read the extracted CSV file, trying default separator
df = pd.read_csv(
    CSV_FILE_NAME,
    encoding='latin1',
    usecols=analysis_columns,
    low_memory=False
    )

print(df.shape)
print(df.head())
```
